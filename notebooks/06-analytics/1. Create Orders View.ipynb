# Create orders View



```sql
CREATE OR REPLACE VIEW olist_catalog.gold.vw_orders 
AS
SELECT 
    oi.order_id,
    oi.order_item_id,
    p.product_id,
    p.product_category_name,
    s.seller_id,
    s.seller_city,
    s.seller_state,
    oi.order_purchase_date_key,
    oi.price,
    oi.freight_value,
    round(oi.price+oi.freight_value,2) as total_price
FROM olist_catalog.gold.fact_order_items oi
JOIN  olist_catalog.gold.dim_products p
ON oi.product_sk==p.product_sk
JOIN olist_catalog.gold.dim_sellers s
ON oi.seller_sk==s.seller_sk
```
