# Transform Order Items Data
1. Filter out invalid rows i.e rows with null order_id or null order_item_id and rows with duplicated order_id & order_item_id
2. check if there are any rows with null product_id and seller_id
3. write the transformed data to the silver table

```python
#Imports
from pyspark.sql.functions import col
```

```python
order_items_df=spark.read.table("olist_catalog.bronze.order_items")
display(order_items_df)
```

### Step1 - Filter out invalid rows

```python
order_items_valid_df = (
    order_items_df.filter(col("order_id").isNotNull() & col("order_item_id").isNotNull())
        .dropDuplicates(subset=["order_id", "order_item_id"])
)
```

### Step2 - check if there are any rows with null product_id and seller_id

```python
display(order_items_valid_df.filter(col("product_id").isNull() | col("seller_id").isNull()))
```

### Step3 - write the transformed data to the silver table

```python
(
    order_items_valid_df.write
        .format("delta")
        .mode("overwrite")
        .saveAsTable("olist_catalog.silver.order_items")
)
```
