creating a function add_ingestion_data which takes the dataframe as input adds the ingestion file and timestamp columns to the dataframe and outputs the dataframe

```python
#Imports
from pyspark.sql.functions import col
from pyspark.sql.functions import current_timestamp
```

```python
def add_ingestion_data(input_df):
    output_df=input_df.withColumns({'source_file':col("_metadata.file_path"),'ingestion_timestamp':current_timestamp()})
    return output_df
```
