```python
#Imports
from pyspark.sql.functions import col
from pyspark.sql.functions import current_timestamp
from pyspark.sql.types import StructType, StructField, StringType, IntegerType,TimestampType, DoubleType
```

```python
source_file="/Volumes/olist_catalog/landing/files/olist_order_reviews_dataset.csv"
table_name = "olist_catalog.bronze.order_reviews"
```

```python
order_reviews_schema = StructType([
    StructField("review_id",StringType()),
    StructField("order_id",StringType()),
    StructField("review_score",IntegerType()),
    StructField("review_comment_title",StringType()),
    StructField("review_comment_message",StringType()),
    StructField("review_creation_date",TimestampType()),
    StructField("review_answer_timestamp",TimestampType())
])
```

```python
order_reviews_df = (
    spark.read
    .format("csv")
    .option("header","True")
    .schema(order_reviews_schema)
    .load(source_file)
    )
```

```python
order_reviews_df_final=add_ingestion_data(order_reviews_df)
```

```python
display(order_reviews_df_final)
```

```python
(
    order_reviews_df_final.write
    .format('delta')
    .mode('overwrite')
    .saveAsTable(table_name)
)
```
