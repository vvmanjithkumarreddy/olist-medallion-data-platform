```python
#Imports
from pyspark.sql.functions import col
from pyspark.sql.functions import current_timestamp
from pyspark.sql.types import StructType, StructField, StringType, IntegerType,TimestampType
```

```python
source_file="/Volumes/olist_catalog/landing/files/olist_products_dataset.csv"
table_name = "olist_catalog.bronze.products"
```

```python
products_schema = StructType([
    StructField("product_id",StringType()),
    StructField("product_category_name",StringType()),
    StructField("product_name_lenght",IntegerType()),
    StructField("product_description_lenght",IntegerType()),
    StructField("product_photos_qty",IntegerType()),
    StructField("product_weight_g",IntegerType()),
    StructField("product_length_cm",IntegerType()),
    StructField("product_height_cm",IntegerType()),
    StructField("product_width_cm",IntegerType())
])
```

```python
products_df = (
    spark.read
    .format("csv")
    .option("header","True")
    .schema(products_schema)
    .load(source_file)
    )
```

```python
products_df_final=add_ingestion_data(products_df)
```

```python
display(products_df_final)
```

```python
(
    products_df_final.write
    .format('delta')
    .mode('overwrite')
    .saveAsTable(table_name)
)
```
