```python
#Imports
from pyspark.sql.functions import col
from pyspark.sql.functions import current_timestamp
from pyspark.sql.types import StructType, StructField, StringType, IntegerType,TimestampType, DoubleType
```

```python
source_file="/Volumes/olist_catalog/landing/files/olist_order_payments_dataset.csv"
table_name = "olist_catalog.bronze.order_payments"
```

```python
order_payments_schema = StructType([
    StructField("order_id",StringType()),
    StructField("payment_sequential",IntegerType()),
    StructField("payment_type",StringType()),
    StructField("payment_installments",IntegerType()),
    StructField("payment_value",DoubleType()),
])
```

```python
order_payments_df = (
    spark.read
    .format("csv")
    .option("header","True")
    .schema(order_payments_schema)
    .load(source_file)
    )
```

```python
order_payments_df_final=add_ingestion_data(order_payments_df)
```

```python
display(order_payments_df_final)
```

```python
(
    order_payments_df_final.write
    .format('delta')
    .mode('overwrite')
    .saveAsTable(table_name)
)
```
