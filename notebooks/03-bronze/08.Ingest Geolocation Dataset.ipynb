```python
#Imports
from pyspark.sql.functions import col
from pyspark.sql.functions import current_timestamp
from pyspark.sql.types import StructType, StructField, StringType, IntegerType,TimestampType, DoubleType
```

```python
source_file="/Volumes/olist_catalog/landing/files/olist_geolocation_dataset.csv"
table_name = "olist_catalog.bronze.geolocation"
```

```python
geolocation_schema = StructType([
    StructField("geolocation_zip_code_prefix",IntegerType()),
    StructField("geolocation_lat",DoubleType()),
    StructField("geolocation_lng",DoubleType()),
    StructField("geolocation_city",StringType()),
    StructField("geolocation_state",StringType())
])
```

```python
geolocation_df = (
    spark.read
    .format("csv")
    .option("header","True")
    .schema(geolocation_schema)
    .load(source_file)
    )
```

```python
geolocation_df_final=add_ingestion_data(geolocation_df)
```

```python
display(geolocation_df_final)
```

```python
(
    geolocation_df_final.write
    .format('delta')
    .mode('overwrite')
    .saveAsTable(table_name)
)
```
